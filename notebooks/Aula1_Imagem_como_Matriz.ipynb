```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color
from google.colab import files
```

## 1. A imagem é uma matriz

Vamos carregar uma imagem e explorar sua estrutura como matriz de números.

```python
# Envie uma foto de folha ou de lavoura
nome = list(files.upload().keys())[0]
img = io.imread(nome)[..., :3]  # descarta o canal alfa, se existir

print("Formato:", img.shape, img.dtype)  # (altura, largura, 3) uint8
print("Pixel (linha 100, coluna 50):", img[100, 50])  # [R G B]
```

## 2. Espaços de cor: RGB, HSV e L*a*b*

```python
hsv = color.rgb2hsv(img)  # H, S e V entre 0 e 1
lab = color.rgb2lab(img)  # L* de 0 a 100; a* e b* negativos e positivos

fig, ax = plt.subplots(1, 4, figsize=(16, 4))
ax[0].imshow(img)
ax[0].set_title("RGB")
ax[1].imshow(hsv[..., 0], cmap="hsv")
ax[1].set_title("H (matiz)")
ax[2].imshow(lab[..., 1], cmap="RdYlGn_r")
ax[2].set_title("a* (verde <-> vermelho)")
ax[3].hist(img[..., 1].ravel(), bins=256, color="green")
ax[3].set_title("Histograma do G")
for a in ax[:3]:
    a.axis("off")
plt.tight_layout()
plt.show()
```

## 3. Quantização: quantos níveis você aguenta perder?

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for i, bits in enumerate([8, 4, 2, 1]):
    passo = 256 // 2**bits
    q = (img // passo) * passo
    axes[i].imshow(q)
    axes[i].set_title(f"{bits} bits = {2**bits} níveis")
    axes[i].axis("off")
    print(f"{bits} bits -> {len(np.unique(q[..., 1]))} tons de verde")

plt.tight_layout()
plt.show()
```

## 4. GSD: quantos centímetros de chão cabem em 1 pixel

O GSD (Ground Sample Distance) é o número mais importante de um voo de drone.

```python
def gsd_cm(altura_m, sensor_mm=13.2, focal_mm=8.8, largura_px=5472):
    """Calcula o GSD em cm/px dado a altura de voo."""
    return altura_m * 100 * sensor_mm / (focal_mm * largura_px)

print("Altura (m)  |  GSD (cm/px)")
print("-" * 25)
for h in [30, 60, 90, 120]:
    print(f"{h:>8} m  |  {gsd_cm(h):.2f} cm/px")
```
